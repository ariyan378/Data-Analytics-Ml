> ### **Lambda Functions**

```python
square = lambda x : x**2
print(square(2))
```

```python
number  = [ 1,2,3,4,5]

# common use with map function
squared = list(map(lambda x : x**2,number))

print(squared)
```

```python
#even number

even = list(filter(lambda x : x %2==0 , number))

print(even)
```

```python
from functools import reduce

product = reduce(lambda acc, x: acc * x, number, 2)
print(product)
```

> ## **File I/O**

```python
#write to file

with open("input.txt" , 'w')as f:
    f.write("Hello\n")

# append to existing file
with open("input.txt" , 'a')as f:
    f.write("Suffer")
```

```python
# read file
with open("input.txt" , 'r') as f:
    content = f.read()
    print(content)
```

```python
# Read line by line
with open("input.txt", "r") as file:
    for line in file:
        print(line.strip())  # strip() removes newline
```

```python
with open("data.txt", "w+") as f:
    f.write("Hello ML")
    f.seek(0)           # back to start for reading
    print(f.read())   # Hello ML
    print(f.tell())   # position after read (end of file)

with open("data.txt", "r") as f:
    chunk = f.read(5)   # first 5 characters
    print(f.tell())     # 5
    f.seek(0)           # jump to beginning again
    print(f.read())     # full file
```
