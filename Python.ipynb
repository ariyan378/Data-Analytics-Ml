> ### **Lambda Functions**

```python
square = lambda x : x**2
print(square(2))
```

```python
number  = [ 1,2,3,4,5]

# common use with map function
squared = list(map(lambda x : x**2,number))

print(squared)
```

```python
#even number

even = list(filter(lambda x : x %2==0 , number))

print(even)
```

```python
from functools import reduce

product = reduce(lambda acc, x: acc * x, number, 2)
print(product)
```
